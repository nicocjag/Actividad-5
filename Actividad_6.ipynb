```python
def filtrar_lecturas_validas(valores_crudos): #[span_0](start_span)[span_0](end_span)
    datos_filtrados = []
    for elemento in valores_crudos:
        if elemento is not None and float(elemento) >= 0:
            datos_filtrados.append(float(elemento))
    return datos_filtrados

def calcular_media(datos):
    tamano = len(datos)
    if tamano == 0:
        return 0.0

    suma_total = sum(datos)
    return suma_total / tamano

def calcular_varianza(datos, media):
    tamano = len(datos)
    if tamano == 0:
        return 0.0

    acumulado_cuadrados = 0.0
    for valor in datos:
        diferencia = valor - media
        acumulado_cuadrados += (diferencia * diferencia)

    return acumulado_cuadrados / tamano

def calcular_desviacion_estandar(varianza):
    return varianza ** 0.5

def calcular_estadisticas_descriptivas(valores_crudos): #[span_1](start_span)[span_1](end_span)
    datos_validos = filtrar_lecturas_validas(valores_crudos)

    if len(datos_validos) == 0:
        return None

    media_calculada = calcular_media(datos_validos)
    varianza_calculada = calcular_varianza(datos_validos, media_calculada)
    desviacion_calculada = calcular_desviacion_estandar(varianza_calculada)

    resultado = {
        "media": media_calculada,
        "varianza": varianza_calculada,
        "desviacion_estandar": desviacion_calculada
    }

    return resultado

datos_XYZ = [12.5, -999, 15.0, None, 18.2, 20.1, -5, 14.8]

res = calcular_estadisticas_descriptivas(datos_XYZ)
print("Resultado:", res)
```
