```python
#Algoritmo selection sort
A = [25.4, 12.8, 35.1, 4.2, 18.5]
for i in range(0, len(A)-1):
    indice_minimo = i
    for j in range(i+1, len(A)):
        if A[j] < A[indice_minimo]:
            indice_minimo = j
    auxiliar = A[i]
    A[i] = A[indice_minimo]
    A[indice_minimo] = auxiliar
print(A)
```
